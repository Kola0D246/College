-> GRAPH

Graph are non linear data structures that consist of a set of vertices (or nodes) and a set of edges that connect pairs of vertices. Graphs can be directed or undirected, weighted or unweighted.

-> Graph Representation in dictionary

object = graph representation stored as dictionary <br>
keys = individual vertecies of graph <br>
value of key = list of vertices connected to key by an edge <br>

```python
#To check if a number is prime number or not

import math
def check_prime(num):
    if type(num) != int or num < 2:
        raise ValueError('num should be positive integer equal to or greater than 2')
    if num <= 3:
        return True
    if num%2 == 0 or num%3 == 0:
        return False
    for x in range(5, math.isqrt(num)+1, 6):
        if num%x == 0 or num%(x+2) == 0:
            return False
    return True

check_prime(25)
```

```python
#To check if number is palindrome
    
def check_palindrome(num):
    num_str = str(num)
    num_str = num_str.replace('.', '')
    num_str = num_str.replace('-', '')
    if num_str.isnumeric():
        return num_str == num_str[::-1]
    raise ValueError('Only numbers are allowed')

check_palindrome('1001')
```

```python
#3. To swap values of two variables inside function

def swap(a, b):
    a, b = b, a
    return a, b

num1, num2 = 1, 2
num1, num2 = swap(num1, num2)
print(num1, num2)
```

```python
#5. To display number of lines in file and file size in bytes

import sys
with open('example.txt', 'r') as file:
    total_lines = 0
    for line in file: 
        total_lines += 1

    print("Total number of lines in file is", total_lines)
    print("Size of file in bytes is", sys.getsizeof(file))
```

```python
#6. To find sum of squares of first N natural numbers

N = int(input("Enter a number"))

if N<1: raise ValueError('N should natural number')

sum = 0
for num in range(1, N+1):
    sum += num**2
print(sum)
```

```python
#7. To find sum of array

arr = [1, 4, 26, 57, -66, 36, 54, 33]

sum = 0
for x in arr:
    if type(arr[0]) != type(x):
        raise TypeError('All elements of array should be of same type')
    sum += x
print(sum)
```

```python
#9. To print even length word from a string

def even_words(string):
    for word in string.split():
        if len(word) % 2 == 0:
            print(word)

string = "I graduated as Bachelor in Technology in Artificial Intelligence and Data Science from Samrat Ashok Technological Institute Vidisha"
even_words(string)
```

```python
#11. To print the ASCII value of a character

ch = input("Enter a character: ")
if len(ch) != 1:
    raise ValueError("Enter only single character")

print(f"ASCII value of {ch} is {ord(ch)}")
```

```python
#12. Find smallest and largest number in list

lst = [2, 5, 53, 63, 1, 46, 8, 43, -54]

smallest = largest = lst[0]
for x in lst:
    if x < smallest: smallest = x
    if x > largest: largest = x

print("Smallest value in lst:", smallest)
print("Largest value in lst:", largest)
```

# IDENTIFIERS

Identifiers are names of variables, functions, class, module or any other objecct in python.

<font size=5> Rules for Identifier </font>

- Can be a combination of uppercase and lowercase letters (case sensitive), digits, or an underscore (_).
- Cannot start with a digit. Only start with letter or underscore
- Cannot contain special symbols like !, #, @, %, $, etc.
- Can be of any length depending on system.
- Keywords in Python cannot be used as identifiers.

```python
myFirstVariable = "Camel Case"

MyFirstVariable = "Pascal Case"

My_First_Variable = "Snake Case 1"
my_first_variable = "Snake Case 2"

print(myFirstVariable)
print(MyFirstVariable)
print(My_First_Variable)
print(my_first_variable)
```

<font size=5> Multiple Words Identifier </font>

- **Camel Case** - Each word except the first starts with capital letter
- **Pascal Case** - Each word starts with capital letter
- **Snake Case** - Each word is separated by a underscore (_) character

### Numeric Datatype

Represents the data that has a numeric value <br>
Three types:- Integer, Float, Complex number

#### Integer Datatype

- keyword `int`
- Represents whole numbers (+ve, -ve, 0) without decimal point
- No limit on size of integer in python
- Default allocated size = 4 bytes

```python
a, b, c, d = 10, -12, 0, 2.0
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))   #This is not int

print(int(d), type(int(d)))     #This is an int. Remove values on right of decimal
```

#### Float Datatype

- keyword `float`
- Represents real numbers (+ve, -ve, 0) with decimal point
- No limit on size of float in python
- Default allocated size = 8 bytes
- can also be expressed in scientific notation

```python
a, b = 10.0, -12.5
print(a, type(a))
print(b, type(b))

print(float(14), type(float(14)))

#in scitific notation -> AeB = A * (10**B)
print(1e3)
print(1e-3)
```

#### Complex number

- Represents numbers in form of `a+bj` where `a` is real part and `bj` is imaginary part.
- `complex(a,b)` function is used to convert a, b into imaginary number.
- `x.real()` function gives real value `a` of any complex number stored in variable x.
- `x.imag()` function gives imaginary value `a` of any complex number stored in variable x.

```python
z = 2+3j
print(z.real)
print(z.imag)
print(type(z))

y = complex(2,3)
print(y)
```

### Dictionary Datatype

- Collection of `<key>: <value>` enclosed inside braces { }

```python
dic = {
    "Name": "Dhruv",
    "Age": 12,
    "Class": "AIADS"
  }

print(dic)
print(type(dic))

for key in dic:
  print(f"{key} -> {dic[key]}")  #This type of print statement is called f-string or formatted string literal

dic2 = {}  #create empty dictionary
```

## Mutable and Immutable

**Immutable Objects** can't be changed after it is created. <br>
example: datatypes like int, float, bool, string, set, Unicode, and tuple.

**Mutable Objects** can be edit in same memory location after they are already created. <br>
example: datatypes like list, dict

```python
#immutable
a = 3
b = a  #now a and b both point to same list in memory
print(a)
print(b)
a = 4   #a and b point to same object but value of a changed and a is immutable so it create a new object in memory and value of b remain same.
print(a)
print(b)
```

```python
#List - mutable
a = [1, 2, 3]
b = a   #now a and b both point to same list in memory
print(a)
print(b)
a += [4]   #this command only concatinate a, but beacuse a and b point to same list and list are mutable so no new list is form, changes affect b also
print(a)
print(b)
```

### Comparison Operator

Compare values on either side of operator:
- if operator correctly say the comparison between both values, then output is `True`
- if operator incorrectly say the comparison between both values, then output is `False`

Operator | Name | Operation | Work
---------|------|-----------|-----
< | lesser than | a<b | `True` if a is lesser than b
\> | greater than | a>b | `True` if a is greater than b
== | equals to | a==b | `True` if a is equals to b
!= | NOT equals to | a!=b  | `True` if a is NOT equals to b
<= | lesser than OR equals to | a<=b | `True` if a is lesser than OR equals to b
\>= | greater than OR equals to | a>=b | `True` if a is greater than OR equals to b

```python
a, b = 10, 20
print(a<b)
print(a>b)
print(a==b)
print(a!=b)
print(a<=b)
print(a>=b)
```

### Logical Operator

Use for decision making

Operator | Work
---------|-----
and | `True` if both statements on either side of operator are `True`
or | `True` if any statement on either side of operator are `True`
not | change `True` -> `False` and `False` -> `True`

```python
a, b = 10, 20
print(a<b and a>b)  #Return false since only one evalue is True
print(a<b or a>b)   #Return false since one evalue is True
print(not(a<b and a>b))   #Change the False to True
```

### Identity Operator

Use for checking if two variable points to same object in memory

Operator | Work
---------|-----
is | `True` if both variable points to same object
is not | `False` if both variable points to same object

```python
a = 10
b = 10
print (a==b)    #a and b are both 10 so True
print(a is b)   #a and b are equal and stored in same memory locations ie, same object so True
print(a is not b)   #opposite of is

print(id(a), id(b))

a += 0.0        #because we add a float value to a, a also become float and hence need a new memory location. (a+=0 will not change memory location)
print(a==b)     #a and b are still equal
print(a is b)   #but now a and b are stored in different memory locations ie, different object so False
print(id(a), id(b))
```

### Membership Operator

Use for checking if left operand (a value or sequence of values) is part of right operand (larger sequence of values)

Operator | Work
---------|-----
in | `True` if left operand is inside right operand
not in | `False` if left operand is inside right operand

```python
a = [1, 2, 3, 4, 5]
print(1 in a)
print(6 in a)
print(1 not in a)
print(6 not in a)
```

# Expression

An expression is a combination of operators and operands that is interpreted to produce some other value.
There can be any number of operator and operand (atleast one of each) in an expression. But there is only one possible output for an expression.

## Types of Expression

1. **Constant expression** - Using constant operands like 2+3, 5*4, etc.
2. **Arithmatic expression** - Perform arithmatic operations like a+b, x/2, etc.
3. **Relational expression** - Do comparison between operands like a<b, 2!=5, etc.
3. **Logical expression** - logical operators (and, or, not) are used.
4. **Bitwise expression** - bitwise operators are used.
5. **Combination Expression** - combination of different types of expression.

### Subscript Operator

To access any element at particular index of sequence data type, we use subscript operator [ ]

syntax: `<var name>[<index of element>]`

````python
str1 = "PROGRAMMING"
print(str1[3])
print(str1[-5])
````

### String Slicing

A technique to extract a substring using subscript operator [ ]

syntax: `<var name>[<start>:<stop>:<step>]`

````python
str1 = "PROGRAMMING"
print(str1[0:3])
print(str1[5:11])
print(str1[7:])     #Leaving stop blank will iterates till the end of list
print(str1[:6])     #Leaving start blank will start iteration from index 0
print(str1[2:9:2])  #Every second elements will be taken in iteration
````

````python
'''Negative indexing means Value of step is negative'''
print(str1[::-1])   #this reverse the characters in str1
````

# Library and Modules

A Python library is a collection of pre-written code modules that help in performing tasks more efficiently. They contain functions, classes, and constants that can be reused to build applications.
- Save time: Avoid reinventing the wheel by using existing code
- Build more complex applications: Use reusable functions and classes to build more robust applications
- Simplify coding: Use libraries to perform tasks like data analysis, machine learning, and web development

***Difference between Module / Package / Library / Framework***

- **Module** <br>
The smallest unit of organization in software.
A piece of code that can be reused or called in a program.
A section of code that can be added as a whole.
Can define functions, classes, and variables.
- **Package** <br>
A collection of modules that can be organized hierarchically.
A namespace that organizes related classes and interfaces.
Similar to a folder in a file directory.
Can contain subpackages.
- **Library** <br>
A collection of modules or packages.
A collection of pre-written code.
A collection of reusable chunks of code.
Can provide a large number of functionalities.
- **Framework** <br>
A set of principles for developing applications.
A collection of modules, packages, and libraries.
Provides a foundation with pre-defined layouts.
Provides an infrastructure that has already figured out the low-level details.
Provides pre-written code libraries, templates, and tools.

## using alises module
Using aliases for modules in Python is a common practice to make code more concise or improve readability. You can assign a shorter name to a module or function by using the as keyword. Here are some examples of how to use aliases for modules:
1. Importing a Module with an Alias:-
You can use an alias to shorten the module name, which is especially useful if the module's name is long or used frequently.

2. 2. Importing Specific Functions or Classes with an Alias:-
If you're importing specific functions or classes from a module, you can also alias them.

3. 3. Aliasing for Libraries with Long Names:-
Some libraries have long names, and using an alias is a great way to shorten the code.

4. 4. Multiple Aliases:-
You can also import multiple modules with different aliases in the same script.

Why Use Aliases?
Readability: Shorten long or frequently used module names.
Avoiding Naming Conflicts: If two modules have similar function names, you can use aliases to differentiate them.
Conventions: Some libraries, like pandas (pd), matplotlib (plt), and numpy (np), are widely known to have these standard aliases.

# Text files in python

It stored a data in a human readable formet. can be opened using a text editor.

opening a file in python-

        variable name = open("file name ","mode")
    ex- file = open("abc.txt","r")


* file modes:-
1. 'r': Read (default mode). Opens the file for reading only. The file must exist.
2. 'w': Write. Opens the file for writing. If the file exists, it will be overwritten.
3. 'a': Append. Opens the file for writing but keeps the existing contents, writing new data at the end.
4. 'b': Binary. Used with other modes to open a file in binary mode (e.g., 'rb', 'wb').
5. 'x': Exclusive creation. Creates a new file but fails if the file already exists.
6. 't': Text. Opens the file in text mode (this is the default and can be omitted).
7. 'r+': Read and write. Opens the file for both reading and writing.
8. 'w+': Write and read. Opens the file for both writing and reading.
9. 'a+': Append and read. Opens the file for both reading and appending.

## CSV

`csv` module provide functions and variables used to manipulate .csv datafile.

CSV (Comma Separated Values) is a plain text file that store data in tabular form where -
- Each row represent a record and separated by a newline
- Each column represents a heading and separated by a comma/ semicolon/ pipe

These files take laest space to store data

<br> Example:
<br> Name, Age, City <-- This is heading
<br> Rudra, 5, Bhopal
<br> Tulse, 6, Vidisha
<br> Saloni, 7, Jabalpur

````python
import csv
with open('data.csv', 'w') as file:     #write .csv file
    writer = csv.writer(file)           #data is declared after file is created in csv module
    writer.writerow(["SN", "Name", "Age", "City"])
    writer.writerow([1, "Rudra", 5, "Bhopal"])
    writer.writerow([2, "Tulse", 6, "Vidisha"])
    writer.writerow([3, "Saloni", 7, "Jabalpur"])
````

````python
with open('data.csv', 'r') as file:     #Read .csv file
    reader = csv.reader(file)
    for row in reader:
        print(row)
````

## JSON file

A JSON file contain `key`:`value` pair organised in a structured way using object { } and arrays [ ]
<br> JSON file are use in web development
<br>
<br> Example of an object/ record/ row in JSON file
```
{
"Name":"Rudra",
"Age":5,
"City":"Bhopal",
"Course":["C","C++","Python","DSA"]
}
```

````python
import json
data = [                        #list include all the objects separated by comma
        {                       #This is row 1
          "Name":"Rudra",
          "Age":5,
          "City":"Bhopal",
          "Course":["C","C++","Python","DSA"]
        },
        {                       #This is row 2
          "Name":"Tulse",
          "Age":6,
          "City":"Vidisha",
          "Course":["C","C++","Python","DSA"]
        },
        {                       #This is row 3
          "Name":"Saloni",
          "Age":7,
          "City":"Jabalpur",
          "Course":["C","C++","Python","DSA"]
        }
]

with open('data.json', 'w') as file:
    json.dump(data, file, indent=4)    #dump(<data>, <json file varaiable>, <other parameters>) is function of json module use to create json file

with open('data.json', 'r') as file:
    data = json.load(file)             #load(<json file variable>) is function of json module use to read json file
    print(data)                        #This print entire data in single line

    for records in data:
      print(records)                   #This print each row of data
````

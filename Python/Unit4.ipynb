# List

A List in Python is a *Mutable*, *Ordered* collection that can store multiple items (called elements) in a single variable and in adjacent memory locations. Following are the key characteristics of a List in Python:

- **Ordered** - All the elements are arranged in a particular order/ sequence.
- **Mutable** - Editing a list does not create a new object in system memory  and changes the data in the same memory location.
- **Hetrogeneous** - Accomodate different datatypes in same list.
- **Indexed** - Each element has a unique position value called index which is use to order the list and to access that particular element. Indexing in list is similar to indexing in String

List in Python vs Array in C

1. List in Python accomodate different datatypes but Array in C allows elements from only one datatype that was assign to array during its declaration.
2. List in Python are extendable infinitly but Array in C can only have specified number of elements in it

<br> List in Python: `list = [1, 2, 'hello', 2+3j, 3.14]`
<br> Array in C: `int array[5] = [2, 4, 5, 23, 4]`

## Accessing List elements

We can access an element or a group of elements from a list using their index number in a subscript operator.
- Retrieving a single element is called Accessing
- Retriving a group of element (a sublist) is called Slicing

```python
colors= ["red", "green", "blue"]
print(colors[2])    #positive indexing
print(colors[-3])   #negative indexing
print(colors[1:3])  #slicing the list
print(colors[::-1]) #reversing the list
print(colors[:])
```

## Modify list

### Update elements

```python
colors[1] = "pink"    #This will change "green" at index 1 to "pink"
print(colors)
```

```python
colors[0], colors[2] = colors[2], colors[0]  #swap elements on index 0 and 2 in list
print(colors)
```

### Adding Element

```python
colors.append("yellow")     # append(<value>) adds the value as element at the end of list
print(colors)
```

```python
colors.insert(3, "blank")   # insert(<index>, <value>) adds the value as element at specified index in list
print(colors)
```

```python
colors.extend(["brown", "white", "pink"]) #add another list at end of the target list (colors)
print(colors)                             #this is use to add more than one element at once
                                          #this can also happen with concetanation operator (+)
```

### Removing element

```python
colors.remove("pink")     #This will remove the first intance of "pink"
print(colors)
```

```python
value = colors.pop(4)     #pop(<index>) removes the element on given index. It returns the remove value
print(value)
print(colors)
colors.pop()              #pop() removes the last element of list by default if no other value is given
print(colors)
```

```python
colors.clear()             #clear() removes all the elements of list
print(colors)
```

```python
del colors          #del removes the list completely. Its an operator not a function
#print(colors)      #Printing list after raise error because it does not exist anymore
```

## Operations on List

```python
lst1 = [2, 3, 4]
lst2 = [5, 6, 7]

print(lst1 + lst2)    #concentation adds two lists
lst1 += lst2          #This works same as lst1.extend(lst2)
print(lst1)

print(lst2 * 3)       #repeatation repeats the elements from the list again and again specified times
```

```python
print(4 in lst2)      #membership operator check if any value is in the list (return True) or not (return False)
print(4 not in lst2)  #reverse membership operator
```

## Multidimentional Lists

```python
#Multi-dimensional lists are the lists within lists
#This is a 2D list. Each list inside the main list is called a record
D2 = [[2, 4, 6, 8, 10], [3, 6, 9, 12, 15], [4, 8, 12, 16, 20]]
print(D2)
```

```python
print(D2[1])      #This is = Print element at index 1 of list D2 - which is a list, so whole list is print
print(D2[2][3])   #This is = Print element at index 3 of the list at index 2 in the list named D2
                  #Similarly, N number of subscript operator will use to acess N dimensional list
```

```python
for record in D2:      #Accessing Records using for loop
    print(record)
```

```python
for record in D2:             #Use nested for loop to access every single element separatly
    for element in record:    #Num of nested for loops = num of dimension of list
        print(element)
```

# Tuple

Tuples are
- comma separated collection of elements enclosed inside parenthesis ( )
- ordered and indexed
- immutable
- hetrogeneous
- allow duplicate elements

```python
tup0 = (1, 2, 3, 'a', 'b', 'c')     #this is a tuple
tup1 = ()                           #this is an empty tuple
tup2 = (10,)                        #this is singleton tuple ie, it has only one elements
tup3 = 12,                          #this is also singleton tuple

#In singleton tuple it is necessary to use comma after the single element or else it wont be stored as a tuple
tup4 = (10)                         #this is not a tuple

print(tup0, type(tup0))
print(tup1, type(tup1))
print(tup2, type(tup2))
print(tup3, type(tup3))
print(tup4, type(tup4))
```

## Tuple packing and unpacking

**Tuple Packing** means assigning multiples values to a single tuple variable, that is tuple declaration or creating a tuple. <br>
`<tuple name> = (<item 1>, <item 2>, ... <item n>)`

**Tuple Unpacking** means extracting values from a tuple and assigning each value to its own variable. *(Number of variables used = length of tuple)* <br>
`<var 1>, <var 2>, ... <var n> = <tuple name>`

```python
#Tuple packing
Student = ("Urmila", "AI", "Vidisha")
print(Student)

#Tuple unpacking
Name, branch, place = Student
print('Name =', Name, '\nBranch =', branch, '\nPlace =', place)
```

## Advantages of Tuple over List

1. Tuple have fixed data assigned to it during time of declaration. The values can <u>not</u> be inserted, edited or deleted from the tuple. *Hence, data is secure and remain unchanged during execution of program*
2. Due to unchangable nature of tuple, it has systamatic and specific memory allocation which is just enough according for data stored in tuple. <br> Comparatively, List have variable memory size because they are mutable and hence take up more memory space than the actual data present in them.
3. As Tuple has fixed data size and fixed memory size, they are more time efficient to work with for larger array length <br>
For example, iterating through tupple is faster than iteration through list of same elments for large number of elements <br>
NOTE: For small number of elements, lists are faster

```python
import sys
tup0 = (1, 2, 3, 4, 5, 6, 7)
lst0 = [1, 2, 3, 4, 5, 6, 7]
print(sys.getsizeof(tup0))
print(sys.getsizeof(lst0))
```

```python
import time

#Time taken for iterating through tuple
begin = time.time()            #store starting time
for x in tup0: pass
stop = time.time()             #store end time
print(f"Total runtime of iterating tuple is {stop - begin}")

#Time taken for iterating through list
begin = time.time()            #store starting time
for x in lst0: pass
stop = time.time()             #store end time
print(f"Total runtime of iterating list is {stop - begin}")
```

```python
import time
tup0 = tuple(range(10000000))
lst0 = list (range(10000000))
#Time taken for iterating through tuple
begin = time.time()            #store starting time
for x in tup0: pass
stop = time.time()             #store end time
print(f"Total runtime of iterating tuple is {stop - begin}")

#Time taken for iterating through list
begin = time.time()            #store starting time
for x in lst0: pass
stop = time.time()             #store end time
print(f"Total runtime of iterating list is {stop - begin}")
```

# Dictionary

A python dictionary is datastructure that stores the value in key:value pair (called item) <br>

<font size=5> Features </font>
1. `Key:Value` pair
2. Key must be unique and immutable
3. Values can be similar and mutable
4. It is ordered
5. It is mutable

## Create a Dictionary

```python
#Inside braces {}
dict1 = {
    "Name": "Koustubh",
    "Age": 20,
    "Place": "Shivpuri"
}
print(dict1)
```

```python
#Create empty dictionary and add element later
dict2 = {}    #empty dictionary
dict2["Name"] = "Utkarsh"
dict2["Age"] = -2
dict2["Place"] = "Jabalpur"
print(dict2)

dict3 = dict()  #empty dictionary
dict3["Name"] = "Ashay"
dict3["Age"] = 19
dict3["Place"] = "Bhopal"
print(dict3)
```

```python
#Create a dictionary using dict() function -> Argument = list of items; Item = tuples of key value pair
dict4 = dict([
    ("Name", "Daksh"),
    ("Age", 20),
    ("Place", "Gwalior")
])
print(dict4)
```

## Access Items of dictionary

Functions such as `keys(), items(), values()` create new temperory object in memory ie, allocate separate in memory and store information there.

### key

```python
dict1 = {"Name": "Koustubh", "Age": 20, "Place": "Shivpuri", "Branch": "AI", "Year": 1}
print(dict1)

#Access keys in dictionary
print(dict1.keys()) #return an object containing all the keys inside a list
print(list(dict1))  #make a list of keys of dictionary
print(tuple(dict1)) #make a tuple of keys of dictionary
print(set(dict1))   #make a set of keys of dictionary

#To access individual key of dictionary at position 1 (like index)
print(list(dict1)[1])
#or
keys = list(dict1)
print(keys[2])
```

### value

```python
#Access all the values of dictionary
print(dict1.values())  #return an object containing all the values inside a list

#Access value of any key in dictionary
print(dict1["Place"])
```

### item

```python
#Access all the values of dictionary
print(dict1.items())  #return an object containing list of all the items as tuples of key value pair
```

## Remove items in dictionary

```python
dict1 = {"Name": "Koustubh", "Age": 20, "Place": "Shivpuri", "Branch": "AI", "Year": 1}
print(dict1)

#pop() - to remove the item of specified key name
dict1.pop("Place")  #return nothing
print(dict1)

#Empty pop() not work in dictionary as dictionary iis not indexed and elements are called by keys

#popitem() - remove last item and return value in that item
print(dict1.popitem(), dict1)

#del keyword to remove item
del dict1["Branch"]
print(dict1)

#remove() not work in dict

#clear to remove everything in dict and leave it empty
dict1.clear()   #return None
print(dict1)

#del whole dict
del dict1
#print(dict1)   #raise error
```

## Dictionary inside dictionary

```python
#dictionary inside a dictionary
super_dict = {
    31 : dict1,
    74 : dict2,
    1 : dict3,
    21 : dict4
}
print(super_dict)     #print everything in single line

#Access values of a dictionary inside a dictionary
print(super_dict[1]["Name"])      #mainDict_name[key that stores nested dictionary][key of desired value of nested dictionary]

#print each item of a dictionary in seperate line
for key, value in super_dict.items():
  print(f"{key} -> {value}")

#print each item of nested dictionary in separate line
for key0 in super_dict:
  print(f"{key0}:")
  for key, value in super_dict[key0].items():
    print(f"  {key} -> {value}")      #add indent in starting of line
```

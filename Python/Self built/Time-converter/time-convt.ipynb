```python
class Timestrap:
    def __init__(self, value, unit):
        self.time = (value, unit)

class time_convt:
    wrt_sec = {
  # 'unit': multiple of second,
    'second': 1,
    'minute': 60,
    'hour': 3600,
    'day': 86400,
    'week': 604800,
    'month-1': 2592000,     #30 days
    'month-2': 2678400,     #31 days
    'month-3': 2419200,     #28 days
    'year': 31536000,       #365 days
    'leap-year': 31622400   #366 days
    }

    month1 = []

    def __init__(self, value, unit):
        self.time = (value, unit)

    def decide(self):
        if self.time[1] in month1:
            raise ValueError(f"Unknown unit: {self.time[1]}")
        return (self.time[0] * self.wrt_sec[self.time[1]], 'second')

    def sec(self):
        if self.time[1] not in self.wrt_sec:
            raise ValueError(f"Unknown unit: {self.time[1]}")
        return (self.time[1] * self.wrt_sec[self.time[1]], 'second')
    
    def minute(self):
        self.sec()
        return (self.time[0] / 60, 'minute')
    
    def hour(self):
        self.sec()
        return (self.time[0] / 3600, 'hour')
    
    def day(self):
        self.sec()
        return (self.time[0] / 86400, 'day')
    
    def week(self):
        self.sec()
        return (self.time[0] / 604800, 'week')
    
    def month(self):
        self.sec()
        return (self.time[0] / 2592000, 'month-1')
    

    

        
```
